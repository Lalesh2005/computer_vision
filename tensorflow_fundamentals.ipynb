## Creating Tensors with tf.constant()

```python
# create a scaler (rank 0 tensor)
scaler = tf.constant(7)
scaler
```

```python
# check number of dimensions of the tensors(ndim stand for number of dimensions)
scaler.ndim
```

```python
# create vector
vector = tf.constant([10,20])
print(vector.ndim)
print(vector)
```

```python
# create matrix
matrix = tf.constant([[10,7],
                      [7,10]
                      ])
print(matrix.ndim)
print(matrix)
```

```python
## define datatype
another_matrix = tf.constant([[10,12],
                             [12,10]]
                             ,dtype=tf.float16
                             )
print(another_matrix)
```

```python
# matrix of more than one dimensons
tensor = tf.constant([[[1,2,3,],
                       [4,5,6]
                       ],
                      [[8,9,10],
                       [11,12,13]
                       ]
                      ])
print(tensor)
```

## Rules of the matrix multiplication
1. the inner dimensions must match
2. The resulting matrix has the shape of the outer dimensions

```python
matrix_1 = tf.constant([[1,2,3,],
                        [4,5,6],
                        [7,8,9]
                        ])
matrix_2 = tf.constant([[1,2],
                        [3,4],
                        [5,6]
                        ])
matrix_1,matrix_2
```

```python
matrix_1 @ matrix_2
```

matrix multiplication is also reffered as dot product
we can perform matrix multiplication using
`tf.matmul`
`tf.tensordot`

```python
x = tf.constant([[1,2],
                 [3,4],
                 [5,6]
                 ])
y = tf.constant([[7,8],
                 [9,10],
                 [11,12]
                 ])
x,y
```

```python
tf.tensordot(tf.transpose(x),y,axes=1)
```

generally we transform rather than reshape

## Aggregating tensors
Aggregating tensors = condensing them from multiple values down to a smaller amount of values  

```python
# Get the absolute values
D = tf.constant([-7,-10])
D
```

```python
tf.abs(D)
```

```python
# Lets go through  the follwing forms of aggregation
# 1. get the minimum
# 2. Get the maximum
# 3. Get the mean of the tensor
# 4. Get the sum of the tensor
```

```python
import numpy as np
```

```python
# creating a random tensor
E = tf.constant(np.random.randint(0,100,size=50))
E
```

```python
tf.size(E),E.shape,E.ndim
```

```python
# find the minimum of the tensor
tf.reduce_min(E)
```

```python
# find the maximum of the tensor
tf.reduce_max(E)
```

```python
# find the sum of the tensor
tf.reduce_sum(E)
```

```python
# find the mean
tf.reduce_mean(E)
```

## find the variance of the tensor

```python
import tensorflow_probability as tfp
tfp.stats.variance(E)
```

```python
tf.math.reduce_std(tf.cast(E,dtype=tf.float32))
```

```python
tf.math.reduce_variance(tf.cast(E,dtype=tf.float32))
```

## Find the positional maximum and minimum.

```python
# create a new tensor for finding positional minimum and maximum
tf.random.set_seed(42)
f = tf.random.uniform(shape=[50])
f
```

```python
# find the postional maximum
tf.argmax(f)
```

```python
# index on our largest value postion
f[tf.argmax(f)]
```

```python
# find the max value of f
tf.reduce_max(f)
```

```python
# Find the postional minimum
tf.argmin(f)
```

```python
# find the min using positional index
f[tf.argmin(f)]
```

## Squeezing a tensor(removing all single dimensions)

```python
tf.random.set_seed(42)
g = tf.constant(tf.random.uniform(shape=[50]),shape=(1,1,1,1,50))
g
```

```python
g.shape
```

```python
# reduces too many ones from the matrix
g_squeezed = tf.squeeze(g)
g_squeezed,g_squeezed.shape
```
